# The paper's results, rebuilt from the saved study values

Every table of *Understand Once, Retrieve Many Times* is generated from compact, hash-verified bundles that live in the repository. This notebook clones the repository and runs that generator on a plain CPU runtime. No dataset, GPU, or model is involved: HR is recomputed from the saved per-query rankings where they exist, every value is checked against the saved study values, and the tables are compared with the paper.

**What this is not:** a rerun of the study. Rerunning needs the four datasets and a multi-GPU vLLM setup; [docs/experiments.md](https://github.com/ZeinabTaghavi/SAADI/blob/main/docs/experiments.md) describes that path.

## Every other table

Ablations, robustness, construction statistics, and comparisons each have a folder with the table in CSV, Markdown, and LaTeX, the generator, the source hashes, and the verification result.


```python
import os

for category in ("main", "comparisons", "ablations", "robustness", "construction"):
    root = os.path.join("results", "tables", category)
    if os.path.isdir(root):
        for study in sorted(os.listdir(root)):
            if os.path.isfile(os.path.join(root, study, "table.md")):
                print(f"{category}/{study}")
```
