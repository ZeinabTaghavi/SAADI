## Every other table

Ablations, robustness, construction statistics, and comparisons each have a folder with the table in CSV, Markdown, and LaTeX, the generator, the source hashes, and the verification result.


```python
import os

for category in ("main", "comparisons", "ablations", "robustness", "construction"):
    root = os.path.join("results", "tables", category)
    if os.path.isdir(root):
        for study in sorted(os.listdir(root)):
            if os.path.isfile(os.path.join(root, study, "table.md")):
                print(f"{category}/{study}")
```
